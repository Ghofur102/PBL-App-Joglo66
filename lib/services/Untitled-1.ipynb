```python
nilai_input = input("Masukkan nilai: ").strip()

try:
    nilai = float(nilai_input)
except ValueError:
    print("Nilai tidak valid! Masukkan angka.")
else:
    if 80 <= nilai <= 100:
        print("Lulus")
    elif 0 <= nilai < 80:
        print("Tidak lulus")
    else:
        print("Nilai tidak valid (harus antara 0 - 100)")
```
